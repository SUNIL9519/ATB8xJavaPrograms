**Python Operators:**

1/ Arithmetic Operators

2/ Assignment Operators

3/ Comparision Operators

4/ Logical Operators

5/ Membership Operators


```python
# Arithmetic Operators (+, -, *, /, %, **) => Used with numeric values to perform mathematical operations

x = 5
y = 8
print(x+y, x-y, x*y, x/y, x%y, x**y)
```

```python
# Assignment Operators (=, +=, -=, *=, /=, %=, **=) => Used to assign values to the variable

x , y = 5 , 8

x += y
print(x)
```

```python
x -= y
print(x) # here x = 13 and y = 8, so output is 5
```

```python
x *= y
print(x)
```

```python
# Comparision Operators (==, !=, <, >, <=, >=) => Used to compare two values
```

```python


print(x == y, x != y, x < y, x > y, x <= y, x >= y)


```

```python
# Logical Operators (==, !=, <, >, <=, >=) => Used to combine conditional statements
# print(x and y , x or y, not y)

(x < y) and (x != y)
```

```python
(x < y) and (x == y)
```

```python
(x < y) or (x != y)
```

```python
(x < y) or (x == y)
```

```python
not ((x < y) or (x == y)) #not means reverse , if true then it will give false
```

```python
# Membership Operators (==, !=, <, >, <=, >=) => Used to test if a  sequence is presented in an object or not

# print (x in y, x not in y)

x = [1,2,3,4,5]
y = 4
y in x
```

```python
y = 6
y in x
```

```python
y not in x
```
